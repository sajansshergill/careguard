# 01 — Explore coverage policies

Synthetic, public-style policy text only. No PHI.
This notebook loads the bundled policies and shows how they are chunked into citable clauses.

```python
from careguard.ingest.chunk import chunk_policy
from careguard.ingest.clean import clean_text
from careguard.ingest.parse import load_policies

docs = load_policies("data/policies")
print(f"{len(docs)} policies")
for d in docs:
    print(d["policy_id"], "—", len(d["text"]), "chars")
```

```python
for d in docs:
    clauses = chunk_policy(d["policy_id"], clean_text(d["text"]))
    print(f"\n{d['policy_id']}: {len(clauses)} clauses")
    for c in clauses:
        print(f"  {c['clause_id']}: {c['text'][:90]}...")
```
