# 03 — Eval analysis

Load `data/eval_report.json` produced by `python -m careguard.eval` and inspect per-case scores.

```python
import json
from pathlib import Path

import pandas as pd

path = Path("data/eval_report.json")
if not path.exists():
    print("No report yet. Run: python -m careguard.eval")
else:
    report = json.loads(path.read_text())
    metrics = report.get("metrics", report)
    display(pd.Series(metrics, name="value"))
    cases = report.get("cases") or []
    if cases:
        df = pd.DataFrame(cases)
        display(df)
        print("mean grounding", df["grounding"].mean())
```
