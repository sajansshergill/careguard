# 02 — Prototype retrieval

Build (or load) the policy index and query it the same way the Retriever agent does.

```python
from pathlib import Path

from careguard.ingest.build_index import main as build_index
from careguard.retrieval.vector_store import VectorStore
from config.settings import settings

if not Path(settings.index_path, "vectors.npy").exists():
    build_index("data/policies/")
store = VectorStore.load(settings.index_path)
print(f"{len(store.meta)} clauses in the index")
```

```python
hits = store.query("MRI lumbar spine after six weeks of physical therapy", k=4)
for h in hits:
    print(f"{h['clause_id']}  score={h['score']:.3f}")
    print(h["text"][:180], "\n")
```
